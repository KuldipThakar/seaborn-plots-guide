```python
import seaborn as sns

# Load the dataset
df = sns.load_dataset("diamonds")
df.head()
```

#1 **Relation Plot**

Scatter plots : Use whenYou want to see the relationship between two numerical variables.

Example Use Case:

Relationship between diamond carat and price.
Comparing height vs. weight of athletes.

```python
#Relational Plots (scatter and line plot )

sns.scatterplot(data=df, x="carat", y="price", hue="cut")  # Scatter plot
```

```python
sns.lineplot(data=df, x="carat", y="price")  # Line plot
```

```python
#Relational Plots (scatter and line plot )

sns.scatterplot(data=df, x="carat", y="price", hue="cut")  # Scatter plot

sns.lineplot(data=df, x="carat", y="price")  # Line plot
```

#2**Catagorical Plots **



```python

sns.boxplot(data=df, x="cut", y="price")  # Box plot

#  Example Use Case:
#Use when: You want to show the distribution of numerical data and identify outliers.
# Comparing salary distribution across job roles.
# Checking spread of test scores across different schools.
```

```python
sns.violinplot(data=df, x="cut", y="price")  # Violin plot

# Use when: You want a better representation of the data distribution (like a box plot + KDE).

#  Example Use Case:

# Comparing income distributions between men and women.
# Visualizing the distribution of house prices.

```

```python
sns.barplot(data=df, x="cut", y="price")  # Bar plot

#  Use when: You want to show the average value of a numerical variable for each category.


#  🔹 Example Use Case:

# Average customer rating for different brands.
# Average sales of different product categories.

```

```python
sns.countplot(data=df, x="cut")  # Count plot

#  Use when: You want to count occurrences of each category.

#  Example Use Case:

# Counting the number of male and female passengers on the Titanic.
# Finding how many diamonds belong to each cut category.
```

```python
sns.pointplot(data=df, x="cut", y="price")

# 📌 Use when: You want to compare mean values of categories with confidence intervals.

# Change in average salary over years.
# Comparison of student performance across different teaching methods.
```

```python
sns.boxplot(data=df, x="cut", y="price")  # Box plot
sns.violinplot(data=df, x="cut", y="price")  # Violin plot
sns.barplot(data=df, x="cut", y="price")  # Bar plot
sns.countplot(data=df, x="cut")  # Count plot
```

#**3: Distribution Plots (Show distribution of a numerical variable)**


```python
sns.histplot(data=df, x="price", bins=30)  # Histogram

 #Use when: You want to see the frequency distribution of numerical data.


# Distribution of exam scores.
# Frequency of house prices in a given range.
```

```python
sns.kdeplot(data=df, x="price")  # KDE plot
# Use when: You want to estimate the probability density of a variable.

# Checking the probability density of salaries.
# Identifying peak values in house price distributions.
```

```python
sns.ecdfplot(data=df, x="price")  # ECDF plot

#  Use when: You want to analyze cumulative distributions.

# Example Use Case:

# Checking how many students scored below a certain mark.
# Finding the proportion of houses priced below $500,000.
```

```python
sns.rugplot(data=df, x="price")

#  Use when: You want to show exact data points along a single axis.
# Showing customer purchase prices.
# Highlighting distribution of age groups in a study.
```

#** Regression plots (SHow Trends and relationship with regression)

```python
sns.regplot(data=df, x="carat", y="price")

# Use when: You want to see linear regression between two variables.
# 🔹 Example Use Case:

# Relationship between advertising budget and sales.
# Connection between house size and price.
```

```python
sns.lmplot(data=df, x="carat", y="price", hue="cut")
# Use when: You want to show regression with categorical grouping.

# 🔹 Example Use Case:

# Regression of weight vs. height for different genders.
# Car price vs. engine size for different fuel types
```

#**5:Matrix Plots(Shows correlation and groupings)**


```python
sns.heatmap(df.corr(numeric_only=True), annot=True)

# 📌 Use when: You want to visualize a correlation matrix.


# 🔹 Example Use Case:

# Checking correlation between stock prices.
# Identifying highly correlated variables in ML datasets.
```

```python
sns.clustermap(df.corr(numeric_only=True), annot=True)

#  Use when: You want to cluster similar data points.

# 🔹 Example Use Case:

# Clustering customer segments based on behavior.
# Grouping similar stock market patterns.
```

#**6. Multi-Plot Grids (Show multiple plots at once)**

```python
sns.pairplot(df[["carat", "price", "depth"]])

# 📌 Use when: You want to visualize relationships between multiple numerical variables.

# Example Use Case:

# Comparing relationships between height, weight, and age.
# Finding trends in various stock attributes.
```

```python
sns.jointplot(data=df, x="carat", y="price", kind="scatter")
#  Use when: You want to see Example Use Case:

# Relationship between age and income.
# Examining sales trends across different store sizes.both scatter + histogram/KDE.
```

#**7. Time-Series Plots**

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load sample dataset
df = sns.load_dataset("fmri")  # Example dataset with time series

# Plot line graph
sns.lineplot(data=df, x="timepoint", y="signal", hue="event")
plt.show()


#  Use when: You want to analyze trends over time.

# Example Use Case:

# Stock market trends over the years.
# Monthly revenue changes for a business.
```
